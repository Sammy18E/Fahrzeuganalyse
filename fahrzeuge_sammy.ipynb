```python
# 1. Frage:


# Gruppieren nach Marke und Mittelwert berechnen
verbrauch_pro_marke = (
    vehicles.groupby("Make")["Annual Petroleum Consumption For Fuel Type1"]
    .mean()
    .reset_index()
    .rename(columns={"Annual Petroleum Consumption For Fuel Type1": "AvgPetroleumConsumption"})
    .sort_values("AvgPetroleumConsumption", ascending=False)
)
```

```python
fig = px.bar(
    verbrauch_pro_marke,
    x="Make",
    y="AvgPetroleumConsumption",
    title="Durchschnittlicher jährlicher Erdölverbrauch pro Automarke",
    labels={"Make": "Automarke", "AvgPetroleumConsumption": "Verbrauch (Gallonen/Jahr)"},
    color="AvgPetroleumConsumption",
    color_continuous_scale="Reds"
)

fig.update_layout(
    xaxis_tickangle=-45,
    template="plotly_white"
)

fig.show()
```

```python
# 2. Frage:


# Wichtige Spalten prüfen & filtern
vehicles_filtered = vehicles[["Make", "Model", "City Mpg For Fuel Type1"]].dropna()

# Duplikate vermeiden durch Kombination von Name + Modell + Verbrauch
vehicles_filtered["Fahrzeug"] = (
    vehicles_filtered["Make"] + " " + vehicles_filtered["Model"] +
    " (" + vehicles_filtered["City Mpg For Fuel Type1"].astype(str) + " MPG)"
)

# Nur eindeutige City MPG-Werte zulassen
vehicles_unique = vehicles_filtered.drop_duplicates(subset=["City Mpg For Fuel Type1"])

# Top 10 effizienteste Fahrzeuge
top10 = vehicles_unique.sort_values("City Mpg For Fuel Type1", ascending=False).head(10)
top10["Effizienzgruppe"] = "Top 10 (effizient)"

# Flop 10 ineffizienteste Fahrzeuge
flop10 = vehicles_unique[vehicles_unique["City Mpg For Fuel Type1"] > 0].sort_values("City Mpg For Fuel Type1").head(10)
flop10["Effizienzgruppe"] = "Flop 10 (ineffizient)"

# Kombinieren
combined = pd.concat([top10, flop10])

# Balkendiagramm erstellen
fig = px.bar(
    combined,
    x="City Mpg For Fuel Type1",
    y="Fahrzeug",
    color="Effizienzgruppe",
    orientation="h",
    title="Top & Flop 10 Fahrzeugmodelle nach City MPG (Stadtverbrauch)",
    labels={"City Mpg For Fuel Type1": "Miles per Gallon (City)", "Fahrzeug": "Fahrzeugmodell"},
    color_discrete_map={
        "Top 10 (effizient)": "green",
        "Flop 10 (ineffizient)": "red"
    }
)

fig.update_layout(
    yaxis=dict(categoryorder="total ascending"),
    template="plotly_white",
    height=700
)

fig.show()
```

```python
# 3. Frage:


# Nur relevante Spalten & Nullwerte entfernen
fuel_df = analyze_df[["Fuel Type1", "Annual Petroleum Consumption For Fuel Type1"]].dropna()

# Gruppieren nach Kraftstofftyp & Ø-Verbrauch berechnen
avg_consumption = (
    fuel_df
    .groupby("Fuel Type1")["Annual Petroleum Consumption For Fuel Type1"]
    .mean()
    .reset_index()
    .rename(columns={"Annual Petroleum Consumption For Fuel Type1": "Ø Verbrauch (Gallonen/Jahr)"})
)

# Sortieren nach Verbrauch
avg_consumption = avg_consumption.sort_values("Ø Verbrauch (Gallonen/Jahr)", ascending=False)

# Plot erstellen
fig = px.bar(
    avg_consumption,
    x="Fuel Type1",
    y="Ø Verbrauch (Gallonen/Jahr)",
    title="Ø jährlicher Erdölverbrauch nach Kraftstofftyp (Fuel Type1)",
    color="Fuel Type1",
    template="plotly_white"
)

fig.update_layout(showlegend=False)
fig.show()
```

```python
# 4. Frage:



# Relevante Spalten & NaNs entfernen
yearly_consumption = vehicles[[
    "Year", "Annual Petroleum Consumption For Fuel Type1"
]].dropna()

# Gruppieren nach Jahr & Mittelwert berechnen
avg_by_year = (
    yearly_consumption
    .groupby("Year")["Annual Petroleum Consumption For Fuel Type1"]
    .mean()
    .reset_index()
    .rename(columns={"Annual Petroleum Consumption For Fuel Type1": "Ø Erdölverbrauch (Gallonen/Jahr)"})
)

# Sortieren nach Jahr (optional, falls nicht chronologisch)
avg_by_year = avg_by_year.sort_values("Year")

# Plot: Liniendiagramm
fig = px.line(
    avg_by_year,
    x="Year",
    y="Ø Erdölverbrauch (Gallonen/Jahr)",
    title="Entwicklung des durchschnittlichen Erdölverbrauchs nach Modelljahr",
    markers=True,
    labels={"Year": "Modelljahr"},
    template="plotly_white"
)

fig.update_traces(line_color="green")
fig.show()
```

```python
# 5. Frage:


# Schritt 1: Spalten auswählen & fehlende Werte entfernen
vehicles_filtered = vehicles[["Vehicle Size Class", "Annual Petroleum Consumption For Fuel Type1"]].dropna()

# Schritt 2: Gruppieren nach Fahrzeugklasse
verbrauch_nach_klasse = (
    vehicles_filtered
    .groupby("Vehicle Size Class")["Annual Petroleum Consumption For Fuel Type1"]
    .mean()
    .reset_index()
    .rename(columns={"Annual Petroleum Consumption For Fuel Type1": "Ø Verbrauch (Gallonen/Jahr)"})
)

# Schritt 3: Nach Verbrauch sortieren
verbrauch_nach_klasse = verbrauch_nach_klasse.sort_values("Ø Verbrauch (Gallonen/Jahr)", ascending=False)

# Schritt 4: Plot erstellen
fig = px.bar(
    verbrauch_nach_klasse,
    x="Ø Verbrauch (Gallonen/Jahr)",
    y="Vehicle Size Class",
    orientation="h",
    title="Ø jährlicher Erdölverbrauch nach Fahrzeugklasse",
    labels={"Vehicle Class": "Fahrzeugklasse"},
    color="Ø Verbrauch (Gallonen/Jahr)",
    color_continuous_scale="Reds"
)

fig.update_layout(template="plotly_white")
fig.show()
```

```python
# 6. Frage:



# Anzahl Fahrzeugmodelle pro Jahr
modelle_pro_jahr = vehicles.groupby("Year")["Model"].nunique().reset_index()
modelle_pro_jahr.columns = ["Modelljahr", "Anzahl Modelle"]

# Liniendiagramm erstellen
fig = px.line(
    modelle_pro_jahr,
    x="Modelljahr",
    y="Anzahl Modelle",
    title="Entwicklung der Anzahl neu eingeführter Fahrzeugmodelle pro Jahr",
    markers=True
)

fig.update_traces(line=dict(color="blue"), marker=dict(size=6))
fig.update_layout(
    xaxis_title="Modelljahr",
    yaxis_title="Anzahl unterschiedlicher Modelle",
    title_font_size=18
)

fig.show()
```

```python
# 7. Frage:


# Nur relevante Spalte betrachten
phev_counts = vehicles["PHEV Blended"].dropna()

# Anzahl Fahrzeuge mit PHEV-Eigenschaften
num_phev = (phev_counts > 0).sum()
num_non_phev = (phev_counts == 0).sum()

# Zusammenfassung in DataFrame
phev_summary = {
    "Typ": ["Plug-in-Hybrid (PHEV)", "Kein Plug-in-Hybrid"],
    "Anzahl": [num_phev, num_non_phev]
}


phev_vehicles = pd.DataFrame(phev_summary)


# Kreisdiagramm erstellen
fig = px.pie(
    phev_vehicles,
    names="Typ",
    values="Anzahl",
    title="Anteil der Fahrzeuge mit Plug-in-Hybrid-Eigenschaften (PHEV)",
    hole=0.3,
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_traces(textinfo='label+percent', pull=[0.05, 0])

fig.show()
```

```python
# 8. Frage:


# Relevante Spalten + NaNs entfernen
df = vehicles[[
    "PHEV Blended",
    "Annual Petroleum Consumption For Fuel Type1"
]].dropna()

# Neue Kategorie-Spalte
df["Typ"] = df["PHEV Blended"].apply(lambda x: "PHEV" if x == 1 else "Konventionell")

# Histogramm erstellen
fig = px.histogram(
    df,
    x="Annual Petroleum Consumption For Fuel Type1",
    color="Typ",
    nbins=40,
    barmode="overlay",
    title="Verteilung des jährlichen Erdölverbrauchs: PHEV vs. Konventionell",
    labels={"Annual Petroleum Consumption For Fuel Type1": "Erdölverbrauch (Gallonen/Jahr)"},
    color_discrete_map={"PHEV": "green", "Konventionell": "gray"}
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Jährlicher Erdölverbrauch (Gallonen)",
    yaxis_title="Anzahl Fahrzeuge",
    legend_title="Fahrzeugtyp"
)

fig.update_traces(opacity=0.6)  # setzt die Transparenz (Deckkraft) der Balken oder Linien auf 60 % (0.6), anstatt der Standard-Deckkraft von 100 % (1.0).

fig.show()
```

```python
# 9. Frage:


# Nur benötigte Spalten + NaNs entfernen
df = vehicles[[
    "Year", "PHEV Blended", "Annual Petroleum Consumption For Fuel Type1"
]].dropna()

# Klassifizieren in PHEV oder Konventionell
df["Typ"] = df["PHEV Blended"].apply(lambda x: "PHEV" if x == 1 else "Konventionell")

# Gruppieren nach Jahr & Typ und Mittelwert berechnen
trend_df = (
    df.groupby(["Year", "Typ"])["Annual Petroleum Consumption For Fuel Type1"]
    .mean()
    .reset_index()
    .rename(columns={"Annual Petroleum Consumption For Fuel Type1": "Ø Verbrauch (Gallonen/Jahr)"})
)

# Liniendiagramm erstellen
fig = px.line(
    trend_df,
    x="Year",
    y="Ø Verbrauch (Gallonen/Jahr)",
    color="Typ",
    markers=True,
    title="Trend: Jährlicher Ø Erdölverbrauch von PHEVs vs. Konventionellen Fahrzeugen",
    color_discrete_map={"PHEV": "green", "Konventionell": "gray"}
)

fig.update_layout(
    template="plotly_white",
    xaxis=dict(dtick=1),
    yaxis_title="Ø Verbrauch (Gallonen/Jahr)",
    legend_title="Fahrzeugtyp"
)

fig.show()
```
